```python
from langchain_community.document_loaders import PyPDFLoader

# Initialize and load your PDF
loader = PyPDFLoader("path/to/your/document.pdf")
documents = loader.load()

print(f"Loaded {len(documents)} page(s).")
```

```python
# Install requirement if needed: pip install sentence-transformers
from sentence_transformers import SentenceTransformer

# Downloads and loads the model automatically (~440 MB)
model = SentenceTransformer("NeuML/pubmedbert-base-embeddings")

# Example usage
texts = [
    "Patient presented with acute myocardial infarction.",
    "Left ventricular ejection fraction is reduced."
]

embeddings = model.encode(texts)
print("Embedding shape:", embeddings.shape)  # Output: (2, 768)
```
